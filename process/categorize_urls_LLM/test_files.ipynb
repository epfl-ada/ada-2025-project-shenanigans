```python
import pandas as pd
from pathlib import Path
import json
```

```python
repo_root = Path.cwd().parent.parent

domains = pd.read_csv(
    repo_root/"dataset"/"sponsoreddomains_unshortened.csv.gz", 
    header=None,  
    names=['domain', 'count']  
)

print(domains.head(50))
print(f"Shape: {domains.shape}")
```

```python
# There are 1.2 million urls. Obviously, not all will be relevant

print(f"All domains: {len(domains):,}")

#  Filter domains that appear >= 10 times
domains_filtered_10 = domains[domains['count'] >= 10].copy()
print(f"Domains with >=10 appearances: {len(domains_filtered_10):,}")

#  Filter domains that appear >= 100 times 
domains_filtered_100  = domains[domains['count'] >= 100].copy()
print(f"Domains with >=100 appearances: {len(domains_filtered_100):,}")

#  Filter domains that appear >= 1000 times 
domains_filtered_1000  = domains[domains['count'] >= 1000].copy()
print(f"Domains with >=1000 appearances: {len(domains_filtered_1000):,}")

#  Filter domains that appear >= 10000 times 
domains_filtered_10000  = domains[domains['count'] >= 10000].copy()
print(f"Domains with >=10000 appearances: {len(domains_filtered_10000):,}")

#  Filter domains that appear >= 100000 times 
domains_filtered_100000  = domains[domains['count'] >= 100000].copy()
print(f"Domains with >=100000 appearances: {len(domains_filtered_100000):,}")
```

```python
total_videos = domains['count'].sum()
print(f"Total videos/URLs: {total_videos:,}")

# For each threshold, show both domain count AND video coverage
for threshold in [10, 100, 1000, 10000, 100000]:
    filtered = domains[domains['count'] >= threshold]
    video_count = filtered['count'].sum()
    pct = (video_count / total_videos) * 100
    print(f">={threshold:,} appearances: {len(filtered):,} domains covering {video_count:,} videos ({pct:.1f}%)")
```
