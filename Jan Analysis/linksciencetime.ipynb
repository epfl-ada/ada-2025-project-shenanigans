```python

import plotly.graph_objects as go
fig = go.Figure()

years = [2005,2006,2007,2008,2009,2010,2011,2012,2013,2014,2015,2016,2017,2018,2019]

for i, site in enumerate(main_sites):
    fig.add_trace(go.Scatter(
        x=years,
        y=site_year[i],
        mode='lines+markers',
        name=site
    ))

fig.update_layout(
    title="Website Mentions per Year",
    xaxis_title="Year",
    yaxis_title="Count",
    template="plotly_white"
)

# Save as interactive HTML
fig.write_html("site_mentions.html")

# Show in notebook (optional)
fig.show()
```
